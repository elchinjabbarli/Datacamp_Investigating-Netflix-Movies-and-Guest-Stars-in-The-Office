## 3. Uygulama Dosyalarının Oluşturulması
Bu adımda mevcut repo'daki kodları Colab diskine yazıyoruz.

```python
# Gerekli klasörleri oluştur
!mkdir -p src

# Dosyaları yaz (Bu kısımları gerçek proje dosyalarından kopyalamalısınız)
# Örnek olarak config.py yazımı:
with open('src/config.py', 'w') as f:
    f.write('''import os
import sys
class Config:
    IS_COLAB = True
    MODEL_DIR = "/content/models"
    UPLOAD_DIR = "/content/uploads"
    OUTPUT_DIR = "/content/output"
    LOG_DIR = "/content/logs"
    TEMPLATE_DIR = "/content/templates"
    MODEL_PATH = "/content/models/phi-3.5-vision-instruct.Q4_K_M.gguf"
    MMPROJ_PATH = "/content/models/phi-3.5-vision-instruct-mmproj.bin"

    @classmethod
    def initialize_env(cls):
        for folder in [cls.UPLOAD_DIR, cls.OUTPUT_DIR, cls.LOG_DIR, cls.TEMPLATE_DIR, cls.MODEL_DIR]:
            os.makedirs(folder, exist_ok=True)
''')
```
